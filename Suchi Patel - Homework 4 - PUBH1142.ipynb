<h2>Task: Create a tuple named primes containing the first seven prime numbers. Then use indexing to return the last three primes numbers in the tuple<h2>

```python
# Create a tuple with the first seven prime numbers
primes = (2, 3, 5, 7, 11, 13, 17)

# Return the last three prime numbers using indexing
last_three = primes[-3:]

print(last_three)
```
