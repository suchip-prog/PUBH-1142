### 15. How do you create a cross-tabulation table that shows the mean Country for each combination of Academic Level (rows) and Salary (columns) in a DataFrame df?

```python
pd.crosstab(
    df["Academic_Level"],
    df["Country"],
    values=df["Age"],
    aggfunc="mean"
)
```
