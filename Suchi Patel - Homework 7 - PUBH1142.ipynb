### 1.Create multiple linear regression model with WCC and CRP as predictors of Lipase.

```python
import statsmodels.api as sm
from patsy import dmatrices

y, X = dmatrices(
    'Mental_Health_Score ~ Avg_Daily_Usage_Hours + Sleep_Hours_Per_Night',
    data=df
)

model = sm.OLS(y, X).fit()
print(model.summary())
```

#### A multiple linear regression model was created to examine whether average daily usage hours and sleep hours per night predict mental health score.

### 2. Comment on the F statistic.

#### The F-statistic is 670.4 with a p-value of 1.50e-163, which is far below 0.05. This indicates that the overall model is statistically significant, meaning that the predictors jointly explain a significant portion of the variation in mental health score.

### 3. Comment on the individual coefficients and their p values.

#### The coefficient for Avg_Daily_Usage_Hours is -0.5666 and is statistically significant (p < 0.001). This suggests that increased social media usage is associated with a decrease in mental health score. The coefficient for Sleep_Hours_Per_Night is 0.1939 and is also statistically significant (p < 0.001). This indicates that more sleep is associated with higher mental health scores. Both predictors are statistically significant and have meaningful relationships with the outcome variable.

### 4. Comment on the R^2 value.

#### The R² value is 0.656, meaning that approximately 65.6% of the variation in mental health score is explained by the model. This suggests a relatively strong model fit.
