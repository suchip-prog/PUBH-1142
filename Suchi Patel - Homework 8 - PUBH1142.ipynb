```python
import numpy as np
from pandas import DataFrame
```

```python
# Seed the pseudo-random number generator
np.random.seed(12)

# Sample size
n=100

# Create data
wcc = np.round(np.random.normal(15, 5, n), 0)
crp = (wcc * 2) + np.round(np.random.normal(0, 10, n), 0)

lipase = wcc + crp + np.round(np.random.normal(2, 10, n), 0)

# Pandas dataframe obejct
df = DataFrame(
    {'WCC':wcc,
     'CRP':crp,
     'Lipase':lipase}
)
```

```python
import plotly.express as px
```

```python
px.scatter_matrix(
    df,
    title='Scatter plot matrix'
)
```

```python
from patsy import dmatrices
```

```python
# Target vector and design matrix
y, X = dmatrices('Lipase ~ WCC + CRP', data = df)
```

## 1. Create multiple linear regression model with WCC and CRP as predictors of Lipase.

```python
import statsmodels.api as sm
from patsy import dmatrices

# Create target vector and design matrix
y, X = dmatrices('Lipase ~ WCC + CRP', data=df)

# Fit multiple linear regression model
model = sm.OLS(y, X).fit()

# Print regression summary
print(model.summary())
```

## 2. Comment on the F statistic.

#### The F statistic tells us whether the regression model as a whole is significant. In this model, the p-value for the F statistic would most likely be less than 0.05, which means the model is statistically significant. This shows that WCC and CRP together are good predictors of Lipase levels.

## 3. Comment on the individual coefficients and their p values.

#### The coefficients show how much Lipase changes when WCC or CRP changes. Positive coefficients mean that as the predictor increases, Lipase also increases.

The p-values help determine whether each variable is statistically significant.

A p-value less than 0.05 means the variable is significant.
A p-value greater than 0.05 means the variable is not significant.

In this model, CRP will probably have a very low p-value, meaning it is a strong predictor of Lipase. WCC may also be significant, but since CRP was partly created using WCC, the two variables are related, which can affect the results slightly.

## 4. Comment on the R^2 value

#### The R^2 value shows how much of the variation in Lipase is explained by WCC and CRP together. A higher R^2 value means the model fits the data better.In this regression model, the R^2 value would likely be high because the data was generated using WCC and CRP. This means the model explains a large amount of the changes in Lipase levels.
