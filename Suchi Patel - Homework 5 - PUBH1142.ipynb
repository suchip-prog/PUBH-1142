### 6. How do you create a heatmap in Seaborn for a variable x and y in a DataFrame df?

```python
heatmap_data = df.pivot_table(values='Addicted_Score', index='Gender', columns='Academic_Level', aggfunc='mean')
sns.heatmap(heatmap_data)
plt.show()
```
